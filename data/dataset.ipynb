# GSM8k

```python
from datasets import load_dataset
import json
import os

print("Loading GSM8K dataset from Hugging Face...")
dataset = load_dataset("openai/gsm8k", "main")

data_dir = "/media/volume/llm/llm_steering_reasoning/data/gsm8k"
os.makedirs(data_dir, exist_ok=True)

for split in ['train', 'test']:
    if split in dataset:
        output_path = os.path.join(data_dir, f"{split}.jsonl")
        with open(output_path, 'w') as f:
            for example in dataset[split]:
                f.write(json.dumps(dict(example)) + '\n')
        print(f"Saved {split} split ({len(dataset[split])} examples) to {output_path}")

print("Done!")
```
